- Objective: generate the CAM topography boundary files (`bnd_topo`: `PHIS`, `LANDFRAC`, `LANDM_COSLAT`, `SGH`, `SGH30`) for the MioMIP2 Phase 2 CORE experiment (15 Ma),
  by smoothing and remapping the MioMIP2 paleogeography onto each target atmosphere grid (`fv19`, `ne30np4`, `ne120np4`).
- Workflow:
  1. Inspect the raw MioMIP2 paleogeography (0.25°).
  2. Configure the target atmosphere grids: SCRIP grid files, present-day reference topo files, and output paths.
  3. Smooth the raw topo per target with a Gaussian filter that treats the ocean as sea level (0 m) and keeps the smoothed field over the ocean, so coastlines become ramps rather than cliffs. The filter width -- its full width at half maximum (FWHM), the distance across the kernel where its weight has dropped to half the peak value -- is calibrated automatically per target so that the slopes of the 15 Ma `PHIS` match those of the present-day reference `PHIS` on the same grid.
  4. Regrid the smoothed topo and the landmask onto each target grid with `ncremap -a patch -g` (ESMF patch recovery; ncremap infers the source grid and generates the weights on the fly).
  5. Derive `PHIS` / `LANDFRAC` / `LANDM_COSLAT` and the baseline `SGH` / `SGH30` (linear regression against each target's present-day reference topo, with a 1.5x inflated slope), and write the `cami_bnd_topo` files.
  6. Derive the rotation-interpolation `SGH` / `SGH30` variant: rotate the present-day reference columns to 15 Ma with GPlately (Zahirovic2022 model), fill the coverage gaps by universal kriging, and write a second set of `cami_bnd_topo` files.
- Notes
  - Topo is smoothed and regridded with the ocean at sea level (0 m), never bathymetry, so each coastal cell gets its cell-mean height and `PHIS` tapers across the coast as in CAM's reference topography. A land-only remap renormalized by the valid weight sum would instead give a nearly-ocean cell the full land height (a cliff at the coast, and a spurious 157 km spike at one ne120np4 column).
  - `SGH` / `SGH30` cannot be derived from the source topo itself (0.25° is far coarser than the GTOPO30 data the reference fields were built from), hence the two reference-based routes in Sections 5 and 6.

## 3. Smooth the raw topo for each target grid

A grid can only carry topographic signal down to the scales it resolves, and CAM's reference topography is smoothed beyond that for numerical stability.
The filter treats the ocean as sea level (0 m, never bathymetry) and keeps the smoothed field over the ocean too.
A coastline therefore becomes a ramp that spills a little out to sea rather than a cliff, as in the reference `PHIS`, where 16% / 12% of the fv19 / ne30np4 ocean columns have `PHIS`/g > 1 m.

**How wide to smooth.** The filter is a Gaussian whose width is set by its full width at half maximum (FWHM), a physical distance in km.
The goal is that the 15 Ma `PHIS` is about as smooth as the present-day reference `PHIS` on the same grid, measured by the ratio of their 99.9th-percentile slopes (`q999`, see the smoothness check at the end of Section 5).
There is no universal FWHM for that: as a multiple of the grid spacing it comes out at ≈ 1.3Δx on fv19, 2.0Δx on ne30np4 and 1.0Δx on ne120np4.
The right width depends on how rough the source topography is, on how each reference file was smoothed (the fv19 reference was remapped from 10′ data in 2005; the SE references use a 16×del2 filter), and on the grid.

What does hold on each grid is a smooth power law. With the current inputs, the `q999` slope ratio (15 Ma / reference) varies with FWHM as:

| FWHM / Nyquist wavelength (2Δx) | 0.5 | 0.75 | 1.0 | 1.25 | 1.5 |
|---|---|---|---|---|---|
| `fv19` (2Δx ≈ 420 km) | 1.21 | 0.95 | 0.78 | 0.66 | 0.56 |
| `ne30np4` (2Δx ≈ 220 km) | 1.41 | 1.20 | 1.00 | 0.84 | 0.72 |
| `ne120np4` (2Δx ≈ 56 km) | 1.00 | 0.85 | 0.75 | 0.68 | 0.63 |

**Rule of thumb.** ratio ∝ FWHM<sup>−α</sup>, with α ≈ 0.7 (fv19), 0.6 (ne30np4) and 0.4 (ne120np4).
To divide the ratio by *r*, multiply the FWHM by *r*<sup>1/α</sup>; since α < 1 that is *r*<sup>1.5</sup> to *r*<sup>2.5</sup>.
For example, bringing a ratio of 1.2 down to 1.0 on fv19 takes about 1.2<sup>1.4</sup> ≈ 1.3× the FWHM.

**Auto-calibration.** The notebook applies this rule itself, so no FWHM has to be chosen by hand:

1. start at the target's Nyquist wavelength (FWHM = 2Δx);
2. smooth, regrid with the saved patch weights, derive `PHIS` as in Section 5, and compute the `q999` slope ratio against the reference;
3. update FWHM ← FWHM · (ratio / target)<sup>1/α</sup>, with α first guessed as 0.6 and then estimated from the last two iterations (a secant step in log–log space);
4. stop once the ratio is within ±0.05 of the target (usually 2–4 iterations).

The target statistic, ratio and tolerance (`calib_stat`, `calib_target`, `calib_tol`) are set at the top of the calibration cell.

```python
from scipy.spatial import cKDTree

# Helpers shared by the FWHM calibration below and the final regrid in Section 4.
grav = 9.80616
r_earth_km = 6371.22  # CAM's Earth radius

# The raw grid is *node*-centered: lat -90..90 and lon -180..180, both endpoints
# inclusive. ncremap infers a SCRIP grid from the source file's lat/lon and
# recognizes the explicit +/-90 rows as polar caps (an FV-style "cap" grid), so
# the poles need no special treatment. The seam does: the 0-360 conversion in
# Section 1 maps both -180 and +180 to lon=180, and that duplicated column would
# infer zero-width cells, so drop it.
ds_src0 = ds_raw.drop_duplicates('lon', keep='first')


def write_src(topo_sm, fpath, note):
    '''Write a regrid source file: the landmask plus the smoothed topo `topo_ocn0`.

    Neither field has missing values: `topo_ocn0` puts the ocean at sea level (0 m),
    never bathymetry, so the remap needs no land-only renormalization and each
    coastal cell gets its cell-mean height, as CAM's PHIS expects.
    '''
    ds_src = ds_src0[['landmask']].copy()
    ds_src['topo_ocn0'] = topo_sm.drop_duplicates('lon', keep='first')
    ds_src['topo_ocn0'].attrs = dict(ds_src0['topo'].attrs)
    ds_src['topo_ocn0'].attrs['smoothing'] = note
    ds_src.to_netcdf(
        fpath,
        encoding={v: {'_FillValue': None} for v in ['topo_ocn0', 'landmask', 'lat', 'lon']},
    )


# The first call per target lets ncremap infer the source grid and call
# ESMF_RegridWeightGen itself (`-a patch -g`), saving the weights to cfg['wgt'];
# later calls (the calibration iterations, then Section 4) reuse them with `-m`.
# Patch recovery gives a smoother, higher-order PHIS remap than bilinear; its
# stencil can over/undershoot near sharp gradients, and both clips in Section 5
# (LANDFRAC.clip(0, 1) and `phis < 0` -> 0) tolerate that.
# `--msk_src=none` matters: NCO otherwise picks up `landmask` as the source grid
# mask, drops every ocean cell from the stencil, and coastal LANDFRAC comes out
# near 1 instead of the true land fraction.
def regrid_topo(cfg, src, out):
    if os.path.exists(cfg['wgt']):
        # a weight file left over from a different source grid would silently misregrid
        with xr.open_dataset(cfg['wgt']) as m:
            if m.sizes['n_a'] != ds_src0.sizes['lat'] * ds_src0.sizes['lon']:
                os.remove(cfg['wgt'])
    if os.path.exists(cfg['wgt']):
        how = f"-m {cfg['wgt']}"
    else:
        how = f"-a patch --msk_src=none -g {cfg['atm_scrip']} -m {cfg['wgt']}"
    cmd = f'''
source $LMOD_ROOT/lmod/init/bash
module --force purge
module load ncarenv/24.12 gcc/12.4.0 mpi-serial/2.5.0 esmf/8.8.1 nco/5.3.1
cd {output_dir}
ncremap -t 1 {how} -v topo_ocn0,landmask {src} {out}
'''
    r = subprocess.run(['bash', '-lc', cmd], capture_output=True, text=True)
    if r.returncode != 0:
        print(r.stdout, r.stderr)
        r.check_returncode()


def topo_slope(ds):
    '''Slope [m/km] of PHIS/g between adjacent columns.

    FV (lat, lon): gradient magnitude, hypot of the lon- and lat-direction slopes,
    with the lon spacing scaled by cos(lat). The lat=+/-90 rows are single points,
    so their lon slope is left NaN.
    SE (ncol): there is no lat/lon neighbor structure, so take the steepest slope
    to the 4 nearest columns, each height difference divided by the great-circle
    distance to that column. It samples only 4 directions, so it can fall slightly
    below the full gradient magnitude.
    '''
    h = ds['PHIS'] / grav
    if 'ncol' not in h.dims:
        lat_rad = np.deg2rad(h['lat'])
        dy = r_earth_km * lat_rad.diff('lat')
        coslat = np.cos(lat_rad).where(np.cos(lat_rad) > 1e-6)
        dx = r_earth_km * coslat * np.deg2rad(h['lon']).diff('lon')
        return np.hypot(h.diff('lon') / dx, h.diff('lat') / dy)
    lo, la = np.deg2rad(ds['lon'].values), np.deg2rad(ds['lat'].values)
    xyz = np.stack([np.cos(la) * np.cos(lo), np.cos(la) * np.sin(lo), np.sin(la)], axis=-1)
    chord, idx = cKDTree(xyz).query(xyz, k=5, workers=-1)
    dist_km = 2 * r_earth_km * np.arcsin(chord[:, 1:] / 2)
    hv = h.values
    return xr.DataArray((np.abs(hv[idx[:, 1:]] - hv[:, None]) / dist_km).max(axis=1), dims='ncol')


def slope_stats(ds):
    '''max / q99 / q999 of the PHIS/g slope [m/km].'''
    slope = topo_slope(ds)
    q = slope.quantile([0.99, 0.999]).values
    return {'max': float(slope.max()), 'q99': q[0], 'q999': q[1]}
```

```python
from scipy.ndimage import gaussian_filter1d


def smooth_topo(topo, landmask, fwhm_km):
    '''Gaussian smoothing of `topo` on the raw regular grid, with the ocean at sea level.

    Ocean cells enter as 0 m (not bathymetry) and the result is kept over the ocean
    too, so a coastline becomes a ramp that spills a little out to sea instead of a
    cliff, as in CAM's reference PHIS.
    '''
    lat = topo['lat'].values
    km_per_deg = 111.32
    dlat_km = km_per_deg * float(np.median(np.diff(lat)))
    dlon_km = km_per_deg * float(np.median(np.diff(topo['lon'].values)))
    # The FWHM is the fine-tuning dial: once the coastal cliffs are gone (below), the
    # q999 slope ratio falls steadily as FWHM grows (ratio ~ FWHM^-alpha), which is
    # what lets the calibration below find each target's FWHM in a few steps.
    sigma_km = fwhm_km / (2 * np.sqrt(2 * np.log(2)))

    # Ocean at sea level -- the change that brought q999 back to ~1. The steepest 0.1%
    # of columns were coastal cliffs (mostly the 15 Ma Antarctic ice margins). A
    # land-only filter never tapers toward the coast, so widening it made fv19 worse
    # (q999 ratio 2.01 -> 2.51 at 1.5x FWHM). Entering the ocean as 0 m ramps heights
    # down to the coast (fv19 2.01 -> 1.18), and keeping the result over the ocean
    # (returned unmasked below) lets the ramp run a little out to sea, as in the
    # reference PHIS (-> 0.78). This only holds because Section 4 regrids without
    # land-only renormalization and Section 5 no longer zeroes PHIS over the ocean.
    h = np.where(landmask.values > 0, topo.values, 0.0)
    h = gaussian_filter1d(h, sigma_km / dlat_km, axis=0, mode='nearest')

    nlon = topo.sizes['lon']
    # Same width in km in both directions: the lon sigma (in grid points) grows as
    # 1/cos(lat) because meridians converge.
    for j in range(lat.size):
        # cap the per-row sigma so near-pole rows do not collapse to the mean of the whole latitude circle
        coslat = max(np.cos(np.deg2rad(lat[j])), 1e-3)
        sigma_lon = min(sigma_km / (dlon_km * coslat), nlon / 6)
        h[j] = gaussian_filter1d(h[j], sigma_lon, mode='wrap')

    # Not masked back to land: the spill over the ocean is part of the coastal ramp.
    return xr.DataArray(h, coords=topo.coords, dims=topo.dims, attrs=topo.attrs)
```

```python
# Auto-calibrate the FWHM per target: the q999 slope ratio of the 15 Ma PHIS to the
# reference PHIS should land within calib_tol of calib_target.
calib_stat, calib_target, calib_tol = 'q999', 1.0, 0.05
calib_max_iter = 8
alpha0 = 0.6  # first guess of the exponent in ratio ~ FWHM^-alpha (0.4-0.7 on these grids)

# nominal grid spacing; the calibration starts at the Nyquist wavelength 2*dx
for tgt, dx in {'fv19': 210.0, 'ne30np4': 110.0, 'ne120np4': 28.0}.items():
    targets[tgt]['dx_km'] = dx


def smoothness_ratio(tgt, fwhm_km):
    '''Smooth at `fwhm_km`, regrid, derive PHIS as in Section 5, and return
    (15 Ma / reference ratio of calib_stat, smoothed source field).'''
    cfg = targets[tgt]
    sm = smooth_topo(ds_raw['topo'], ds_raw['landmask'], fwhm_km)
    src = os.path.join(output_dir, f'calib_src_{tgt}.nc')
    out = os.path.join(output_dir, f'calib_out_{tgt}.nc')
    write_src(sm, src, note=f'calibration trial, FWHM {fwhm_km:.0f} km')
    regrid_topo(cfg, src, out)
    with xr.open_dataset(out) as ds:
        ds = ds.reset_coords().load()
    ds['PHIS'] = (ds['topo_ocn0'] * grav).clip(min=0)
    for f in [src, out]:
        os.remove(f)
    return slope_stats(ds)[calib_stat] / cfg['ref_stats'][calib_stat], sm


topo_smooth = {}
for tgt, cfg in targets.items():
    with xr.open_dataset(cfg['ref_topo']) as ds:
        cfg['ref_stats'] = slope_stats(ds.load())

    # keep the search within a sane range of the grid spacing
    fwhm_lo, fwhm_hi = 0.25 * cfg['dx_km'], 8 * cfg['dx_km']
    fwhm, alpha, hist = 2 * cfg['dx_km'], alpha0, []
    for it in range(calib_max_iter):
        ratio, sm = smoothness_ratio(tgt, fwhm)
        hist.append((fwhm, ratio, sm))
        print(f'{tgt}: iter {it}  FWHM {fwhm:6.1f} km  {calib_stat} ratio {ratio:.3f}')
        if abs(ratio - calib_target) <= calib_tol:
            break
        if len(hist) >= 2:
            # secant in log-log space: the local exponent from the last two trials
            (f0, r0, _), (f1, r1, _) = hist[-2], hist[-1]
            alpha = float(np.clip(-np.log(r1 / r0) / np.log(f1 / f0), 0.2, 1.5))
        fwhm_new = float(np.clip(round(fwhm * (ratio / calib_target) ** (1 / alpha), 1), fwhm_lo, fwhm_hi))
        if fwhm_new == fwhm:
            print(f'{tgt}: FWHM stuck at {fwhm:.1f} km (search bound or step below 0.1 km)')
            break
        fwhm = fwhm_new
    else:
        print(f'{tgt}: not within +/-{calib_tol} after {calib_max_iter} iterations; using the closest trial')

    fwhm, ratio, sm = min(hist, key=lambda t: abs(t[1] - calib_target))
    cfg['fwhm_km'], cfg['calib_ratio'] = fwhm, ratio
    topo_smooth[tgt] = sm
    print(f"--> {tgt}: FWHM {fwhm:.1f} km = {fwhm / cfg['dx_km']:.2f} dx "
          f"({fwhm / (2 * cfg['dx_km']):.2f} x Nyquist), {calib_stat} ratio {ratio:.3f}\n")

topo_land_raw = ds_raw['topo'].where(ds_raw['landmask'] > 0)
for tgt, cfg in targets.items():
    diff = topo_smooth[tgt] - topo_land_raw  # NaN over ocean, so land only
    spill = topo_smooth[tgt].where(ds_raw['landmask'] == 0)
    print(f"{tgt}: FWHM {cfg['fwhm_km']:.0f} km; land-mean shift {float(diff.mean()):+.1f} m, "
          f'RMS {float((diff ** 2).mean()) ** 0.5:.0f} m, |diff| max {float(abs(diff).max()):.0f} m; '
          f'max spill over ocean {float(spill.max()):.0f} m')
```

## 5. Derive `PHIS`, `LANDFRAC`, `LANDM_COSLAT`, `SGH`, `SGH30` (baseline)

The is our baseline: SGH / SGH30 are directly derived from present day at the same target grid.
A 1.5x inflation is applied to compensate for the source topo already being smoother than the GTOP030 data the reference SGH was built from.
The Section 3 smoothing makes the source smoother still, which that inflation now also has to cover.

```python
for tgt, cfg in targets.items():
    with xr.open_dataset(cfg['topo_tmp']) as ds:
        ds = ds.reset_coords()

        # Get LANDFRAC from the regridded landmask
        landfrac = ds['landmask'].astype('float64').clip(0, 1)
        landfrac.attrs = {'long_name': 'gridbox land fraction'}
        ds['LANDFRAC'] = landfrac

        # Compute LANDM_COSLAT without smoothing (not used by CAM5/6)
        landm_coslat = landfrac * np.cos(np.deg2rad(ds['lat']))
        landm_coslat.attrs = {
            'long_name': 'land ocean transition mask: ocean (0), continent (1), transition (0-1)',
            'NOTE1': 'This field has not been smoothed',
            'NOTE2': 'This field is not used by CAM5/6',
        }
        ds['LANDM_COSLAT'] = landm_coslat

        # Compute PHIS
        phis = ds['topo_ocn0'].astype('float64') * grav
        phis.attrs = {
            'long_name': 'surface geopotential',
            'units': 'm2/s2',
            'from_hires': 'false',
        }
        phis = xr.where(phis < 0, 0, phis)
        ds['PHIS'] = phis

        # Fit PHIS -> SGH / SGH30 from this target's resolution-matched reference file.
        # Select land columns through a flat numpy mask so both layouts work:
        # the SE references are 1-D on `ncol`, the FV reference is 2-D on (lat, lon).
        ds_ref = xr.open_dataset(cfg['ref_topo'])
        ref_land = ds_ref['LANDFRAC'].values > 0
        phis_ref = ds_ref['PHIS'].values[ref_land]
        sgh_ref = ds_ref['SGH'].values[ref_land]
        sgh30_ref = ds_ref['SGH30'].values[ref_land]

        phis_to_sgh = np.polyfit(phis_ref, sgh_ref, 1)
        phis_to_sgh30 = np.polyfit(phis_ref, sgh30_ref, 1)

        # Compute SGH
        # Artificially increase slope by 50% to account for the coarse/smooth source topo.
        sgh = phis * phis_to_sgh[0] * 1.5 + phis_to_sgh[1]
        sgh.attrs = {
            'long_name': 'standard deviation of 10-min elevations',
            'units': 'm',
            'creation': f"linear regression against {os.path.basename(cfg['ref_topo'])}, 50% larger slope",
        }
        ds['SGH'] = sgh.astype('float64')

        # Compute SGH30
        # Artificially increase slope by 50% to account for the coarse/smooth source topo.
        sgh30 = phis * phis_to_sgh30[0] * 1.5 + phis_to_sgh30[1]
        sgh30.attrs = {
            'long_name': 'standard deviation of elevation from 30s to 10m',
            'units': 'm',
            'creation': f"linear regression against {os.path.basename(cfg['ref_topo'])}, 50% larger slope",
        }
        ds['SGH30'] = sgh30

        # Drop variables that are not needed. A keep-list, so the FV extras
        # (gw, lat_bnds, lon_bnds, area) go the same way as the SE vertices.
        keep = ['lat', 'lon', 'LANDFRAC', 'LANDM_COSLAT', 'PHIS', 'SGH', 'SGH30']
        ds = ds.drop_vars([v for v in ds.data_vars if v not in keep])

        # Set missing value
        for da in ds.data_vars.values():
            da.attrs['missing_value'] = 1.e+36
            da.attrs['_FillValue'] = 1.e+36

        ds.attrs['author'] = 'Feng Zhu (fengzhu@ucar.edu)'

        if os.path.exists(cfg['topo_out']):
            os.remove(cfg['topo_out'])
        ds.to_netcdf(cfg['topo_out'], format='NETCDF3_64BIT')

    print(f'--> {tgt}: {cfg["topo_out"]}')
```

### Check the smoothness of `PHIS`

The present-day reference `PHIS` on each target grid (built from GTOPO30 and already smoothed for CAM; also what CESM uses for preindustrial runs) is a fair yardstick for how rough the topography on that grid may be.
If the Section 3 smoothing works, the 15 Ma `PHIS` should have slopes of the same order (ratio ~1); a much larger ratio would point to grid-scale noise that the dynamical core would have to damp.
The Section 3 FWHM calibration uses the same slope statistics (`slope_stats`).
Slopes are in m/km, so they can also be compared across grids: a finer grid resolves steeper terrain.

How to read the three statistics:

| statistic | what it measures | role |
|---|---|---|
| `q999` | the steepest 0.1% of columns (≈ 14 / 49 / 778 on fv19 / ne30np4 / ne120np4) | **primary target**: this tail is what drives grid-scale noise, spurious vertical motion and stability limits, yet it is a statistic over several columns rather than one |

| `q99` | the steepest 1%, mostly ordinary mountain slopes | **secondary check**: flags over-smoothing (a ratio well below 1) more than stability problems |

#### Power spectra

The slope statistics above sample only the steepest columns. A power spectrum shows the whole range of scales instead, as in Fig. 1 of [Lauritzen et al. (2015)](https://gmd.copernicus.org/articles/8/3975/2015/), which plots the spectral power of surface elevation against total wavenumber *K* on log-log axes for raw and smoothed CAM topography.
Here each `PHIS`/g (ocean at 0 m) goes onto a common 720 × 1440 Gaussian grid (0.25°), and its spherical-harmonic power per *K* is computed with an exact Gauss-Legendre transform, so the power summed over *K* equals the global mean of h².
The vertical line marks each grid's cutoff *K*<sub>c</sub> = π*R*/Δx (wavelength 2Δx). The target-grid curves stop there, since beyond it they would only show the interpolation onto the Gaussian grid.

How to read it:

- the raw 15 Ma source (black) falls off roughly as *K*<sup>−2</sup>, like Earth's topography in Lauritzen et al.;
- smoothing for CAM pulls the target-grid curves below it well before the cutoff;
- if the 15 Ma `PHIS` (red) tracks the reference (blue) near *K*<sub>c</sub>, the two are smoothed to a similar degree there. A red curve well above the blue one in the last octave (*K*<sub>c</sub>/2 to *K*<sub>c</sub>) would mean extra grid-scale energy.

The table gives the 15 Ma / reference power ratio in three bands. At low *K* the ratio reflects the different geographies (15 Ma vs present day), not the smoothing.

```python
# Power spectra of PHIS/g on each target grid, after Fig. 1 of Lauritzen et al.
# (2015, GMD 8, 3975): spectral power per total wavenumber K on log-log axes,
# against the raw source and a K^-2 slope.
from scipy.interpolate import RegularGridInterpolator
from scipy.spatial import ConvexHull

# Every field first goes onto one Gaussian grid. Gauss-Legendre latitudes make
# the Legendre quadrature exact up to degree nlat - 1, so 720 rows resolve
# K <= 719, which covers the 0.25 deg raw source.
spec_nlat = 720
spec_lmax = spec_nlat - 1
spec_mu, spec_w = np.polynomial.legendre.leggauss(spec_nlat)  # mu = sin(lat), S -> N
spec_lat = np.rad2deg(np.arcsin(spec_mu))
spec_lon = np.arange(2 * spec_nlat) * 180 / spec_nlat
_LAT, _LON = np.meshgrid(spec_lat, spec_lon, indexing='ij')


def _xyz(lon, lat):
    lo, la = np.deg2rad(lon), np.deg2rad(lat)
    return np.stack([np.cos(la) * np.cos(lo), np.cos(la) * np.sin(lo), np.sin(la)], axis=-1)


def to_gauss_ll(da):
    '''Bilinear from a regular (lat, lon) grid (raw 0.25 deg, smoothed source, fv19).'''
    da = da.drop_duplicates('lon').sortby('lon')
    lat, lon, v = da['lat'].values, da['lon'].values, da.values
    # pad one column on each side so the lon seam interpolates across
    lonp = np.concatenate([lon[-1:] - 360, lon, lon[:1] + 360])
    vp = np.concatenate([v[:, -1:], v, v[:, :1]], axis=1)
    return RegularGridInterpolator((lat, lonp), vp)((_LAT, _LON))


def to_gauss_se(vals, lon, lat):
    '''Linear interpolation from SE columns on the spherical Delaunay triangulation.

    The convex hull of the columns on the unit sphere is their spherical Delaunay
    triangulation; each Gaussian point takes the barycentric weights of the
    triangle it falls in. This stands in for ESMF bilinear (which segfaults on
    ne120np4 -> this grid) and agrees with it on ne30np4 to a few % in power up to
    the grid cutoff.
    '''
    p = _xyz(lon, lat)
    tri = ConvexHull(p).simplices
    Tinv = np.linalg.inv(np.transpose(p[tri], (0, 2, 1)))  # columns = triangle vertices
    q = _xyz(_LON.ravel(), _LAT.ravel())
    # candidate triangles: the k nearest by centroid; keep the first that contains q
    k = 12
    _, cand = cKDTree(p[tri].mean(axis=1)).query(q, k=k, workers=-1)
    idx, lam = np.full(q.shape[0], -1), np.zeros((q.shape[0], 3))
    for c in range(k):
        todo = np.flatnonzero(idx < 0)
        t = cand[todo, c]
        l = np.einsum('nij,nj->ni', Tinv[t], q[todo])
        ok = (l >= -1e-12).all(axis=1)
        idx[todo[ok]], lam[todo[ok]] = t[ok], l[ok]
    assert (idx >= 0).all(), f'{(idx < 0).sum()} Gaussian points not located'
    lam /= lam.sum(axis=1, keepdims=True)
    return (vals[tri[idx]] * lam).sum(axis=1).reshape(_LAT.shape)


def sh_power(h):
    '''Spherical-harmonic power per total wavenumber K = 0..spec_lmax [m^2].

    `h` is (field, lat, lon) on the Gaussian grid. Orthonormal harmonics, scaled
    so that the sum over K equals the global mean of h^2 (Parseval). An FFT in
    lon, then a Legendre quadrature in lat with the fully normalized associated
    Legendre functions, recursed in K for all orders m at once.
    '''
    h = np.asarray(h, dtype='float64')
    nf, nlat, nlon = h.shape
    L, x = spec_lmax, spec_mu
    F = np.fft.rfft(h, axis=-1)[..., :L + 1] / nlon
    F = np.transpose(F, (0, 2, 1)) * spec_w          # (field, m, lat), quadrature weights folded in

    Pmm = np.empty((L + 1, nlat))                     # sectoral P_mm
    Pmm[0] = 1 / np.sqrt(4 * np.pi)
    for m in range(1, L + 1):
        Pmm[m] = np.sqrt((2 * m + 1) / (2 * m)) * np.sqrt(1 - x ** 2) * Pmm[m - 1]

    power = np.zeros((nf, L + 1))
    P1, P2 = np.zeros((L + 1, nlat)), np.zeros((L + 1, nlat))  # P_{K-1,m}, P_{K-2,m}
    for K in range(L + 1):
        P = np.zeros((L + 1, nlat))
        P[K] = Pmm[K]
        if K >= 1:
            P[K - 1] = np.sqrt(2 * K + 1) * x * Pmm[K - 1]
        if K >= 2:
            m = np.arange(K - 1)[:, None]
            a = np.sqrt((4 * K ** 2 - 1) / (K ** 2 - m ** 2))
            b = np.sqrt(((K - 1) ** 2 - m ** 2) / (4 * (K - 1) ** 2 - 1))
            P[:K - 1] = a * (x * P1[:K - 1] - b * P2[:K - 1])
        c2 = np.abs(2 * np.pi * np.einsum('fmj,mj->fm', F[:, :K + 1], P[:K + 1])) ** 2
        power[:, K] = (c2[:, 0] + 2 * c2[:, 1:].sum(axis=1)) / (4 * np.pi)  # m and -m
        P2, P1 = P1, P
    return power


# Ocean at 0 m throughout, as in the PHIS panels above.
spec_fields = {('raw', 'raw'): to_gauss_ll(ds_raw['topo'].where(ds_raw['landmask'] > 0, 0))}
for tgt, cfg in targets.items():
    spec_fields[(tgt, 'smoothed source')] = to_gauss_ll(topo_smooth[tgt])
    for label, fpath in [('reference (PD)', cfg['ref_topo']), ('15 Ma', cfg['topo_out'])]:
        with xr.open_dataset(fpath) as ds:
            ds = ds.reset_coords().load()  # SE refs carry lat/lon as data variables
        h = ds['PHIS'] / grav
        spec_fields[(tgt, label)] = (to_gauss_se(h.values, ds['lon'].values, ds['lat'].values)
                                     if 'ncol' in h.dims else to_gauss_ll(h))

spec_power = dict(zip(spec_fields, sh_power(np.stack(list(spec_fields.values())))))
K = np.arange(spec_lmax + 1)

# Grid cutoff: the shortest wavelength a grid carries is 2 dx, i.e. K = pi R / dx.
for cfg in targets.values():
    cfg['K_nyq'] = int(np.pi * r_earth_km / cfg['dx_km'])

fig, axes = plt.subplots(1, len(targets), figsize=(4.6 * len(targets), 4.4),
                         sharey=True, constrained_layout=True)
styles = {
    'smoothed source': dict(color='0.55', lw=1, ls='--'),
    'reference (PD)': dict(color='C0', lw=1.6),
    '15 Ma': dict(color='C3', lw=1.6),
}
for ax, (tgt, cfg) in zip(axes, targets.items()):
    k = K[1:]
    ax.loglog(k, spec_power[('raw', 'raw')][1:], color='k', lw=1, label='raw 15 Ma (0.25°)')
    for label, st in styles.items():
        # beyond the cutoff the target-grid curves only show the interpolation, so cut them there
        kmax = spec_lmax if label == 'smoothed source' else cfg['K_nyq']
        ax.loglog(k[k <= kmax], spec_power[(tgt, label)][1:][k <= kmax], label=label, **st)
    # K^-2 slope, anchored to the raw spectrum at K = 10
    ax.loglog(k, spec_power[('raw', 'raw')][10] * (k / 10.) ** -2, color='k', lw=0.8, ls=':', label=r'$K^{-2}$')
    ax.axvline(cfg['K_nyq'], color='0.3', lw=0.8, ls='-.')
    ax.text(cfg['K_nyq'], 0.97, f" 2Δx\n K={cfg['K_nyq']}", transform=ax.get_xaxis_transform(),
            va='top', ha='right', fontsize=9)
    ax.set_title(f"{tgt} (FWHM {cfg['fwhm_km']:.0f} km)")
    ax.set_xlabel('total wavenumber K')
    ax.set_xlim(1, spec_lmax)
    ax.set_ylim(1e-3, 1e5)
axes[0].set_ylabel('power of PHIS/g [m$^2$]')
axes[0].legend(fontsize=9, loc='lower left')

# Quantitative check: 15 Ma / reference power summed over three bands below the
# cutoff. The top band (the shortest resolved scales) is where grid-scale noise
# would show up as a ratio well above 1.
rows = []
for tgt, cfg in targets.items():
    kn = cfg['K_nyq']
    bands = {'K < Kc/4': (1, kn // 4), 'Kc/4 - Kc/2': (kn // 4, kn // 2), 'Kc/2 - Kc': (kn // 2, kn + 1)}
    rows.append(dict(target=tgt, K_c=kn, **{
        b: spec_power[(tgt, '15 Ma')][k0:k1].sum() / spec_power[(tgt, 'reference (PD)')][k0:k1].sum()
        for b, (k0, k1) in bands.items()}))
print('15 Ma / reference: PHIS/g power summed over each band (K_c = grid cutoff)')
print(pd.DataFrame(rows).set_index('target').round(2))
```

## 6. Rotation-Interpolation based SGH / SGH30

Section 5 derives `SGH` / `SGH30` as a linear function of `PHIS`, so the subgrid
roughness is smooth by construction and carries no information about *where* the
terrain actually was. Here we take the alternative route to
carry a present-day field to its palaeo position with the plate model, then fill
whatever gaps the rotation leaves.

Applied to roughness, per target grid:

1. take present-day `SGH` / `SGH30` from that target's **own** reference topo,
   masked to present-day land;
2. rotate those columns to 15 Ma with the Zahirovic et al. (2022) model via
   `gplately.Points` (anchor plate 701701, the palaeomagnetic frame);
3. give each Miocene column the nearest rotated column within one grid spacing —
   the ~5–7% with nothing that close are gaps, filled three ways: left blank,
   nearest neighbour (`sit.fast_interpolate`), and universal kriging with a
   spherical variogram (`sit.slow_interpolate`).

```python
import time
from concurrent.futures import ProcessPoolExecutor
from scipy.spatial import cKDTree
from pykrige.uk import UniversalKriging
import gplately
import plate_model_manager

recon_time = 15  # Ma -- matches the MioMIP2 Phase 2 CORE reconstruction

# ~85 MB of rotations / static polygons / topologies, so keep it out of the repo.
# The first run downloads it from the Plate Model Manager servers; after that the
# readonly branch picks up the local copy.
plate_model_dir = '../../../../utils/platemodels'
os.makedirs(plate_model_dir, exist_ok=True)
try:
    Z22_pmm = plate_model_manager.PlateModel(
        'Zahirovic2022', data_dir=plate_model_dir, readonly=True)
except Exception:
    Z22_pmm = plate_model_manager.PlateModelManager().get_model(
        'Zahirovic2022', data_dir=plate_model_dir)

Z22_static_polygons = Z22_pmm.get_static_polygons()
# anchor_plate_id=701701 is the palaeomagnetic reference frame
Z22_model = gplately.PlateReconstruction(
    Z22_pmm.get_rotation_model(),
    Z22_pmm.get_topologies(),
    Z22_static_polygons,
    anchor_plate_id=701701,
)

# Only needed for the 15 Ma shoreline overlay in the comparison figure -- the
# rotation below never touches a regular grid.
ds_raw025 = xr.open_dataset(raw025_fpath)

# A target column counts as covered if a rotated present-day column lands within
# this many grid spacings of it. 1.0 is the natural reading ("within one cell").
# Sensitivity at ne120np4: 0.75 -> 21.6% gaps, 1.0 -> 7.3%, 1.5 -> 5.3%, 2.0 ->
# 4.5%. The steep drop below 1.0 is sampling noise; the flat tail above 1.5 is
# genuine new Miocene land that no present-day parcel maps onto.
gap_radius_factor = 1.0

# Neighbours per kriging solve. The solve is O(k^3), and once the variogram is
# fixed (see `fit_variogram`) accuracy is flat in k: on held-out columns, k=100 /
# 50 / 25 gave MAE 50.7 / 50.6 / 50.5 m at ne30np4 and 23.8 / 23.6 / 23.2 m at
# ne120np4. 50 keeps comfortable margin over the 3 drift coefficients.
krig_spacing = 50


def _xyz(lon, lat):
    '''Unit-sphere Cartesian coords, so nearest-neighbour searches don't care
    about the lon seam or converging meridians.'''
    lo, la = np.deg2rad(lon), np.deg2rad(lat)
    return np.stack([np.cos(la) * np.cos(lo), np.cos(la) * np.sin(lo), np.sin(la)], axis=-1)


def grid_spacing_deg(lon, lat):
    '''Median nearest-neighbour separation of an unstructured grid, in degrees.'''
    d, _ = cKDTree(_xyz(lon, lat)).query(_xyz(lon, lat), k=2, workers=-1)
    return np.rad2deg(2 * np.arcsin(np.median(d[:, 1]) / 2))


def column_lonlat(ds):
    '''Per-column lon/lat as flat 1-D arrays, for either layout: SE files carry
    one value per column already; FV files carry 1-D axes that mesh out.'''
    if 'ncol' in ds['lon'].dims:  # SE
        return ds['lon'].values, ds['lat'].values
    lon2, lat2 = np.meshgrid(ds['lon'].values, ds['lat'].values)
    return lon2.ravel(), lat2.ravel()


def rotate_columns(ds_src, names, src_land, dst_lon, dst_lat, radius_deg):
    '''Rotate present-day source columns to `recon_time`, sample onto dst columns.

    The SE grids do not move -- only the terrain does -- so this goes straight
    from one set of `ncol` coordinates to another. No regular grid is involved:
    the reference notebook needs one because `gplately.Raster` demands it, but
    `gplately.Points` takes the column lon/lat from the topo file as they are.

    Rotating moves *coordinates*, so one call serves every field in `names`.
    Returns the sampled fields plus the coverage mask (False = gap to fill).
    '''
    src_lon, src_lat = column_lonlat(ds_src)     # FV refs arrive 2-D; work on flat columns
    src_land = np.asarray(src_land).ravel()
    lon_pd = ((src_lon[src_land] + 180) % 360) - 180  # gplately wants -180..180
    lat_pd = src_lat[src_land]

    # `remove_unreconstructable_points=False` is required, not cosmetic: with True
    # the indices returned below refer to gplately's trimmed internal array rather
    # than to `lon_pd`/`lat_pd`, so `values[keep]` would be silently misaligned.
    pts = gplately.Points(Z22_model, lon_pd, lat_pd,
                          remove_unreconstructable_points=False)
    rlon, rlat, keep = pts.reconstruct(
        recon_time, return_array=True, return_point_indices=True)
    assert np.allclose(pts.lons[keep], lon_pd[keep]), 'point indices are misaligned'

    cutoff = 2 * np.sin(np.deg2rad(radius_deg) / 2)  # degrees -> chord length
    dist, j = cKDTree(_xyz(rlon, rlat)).query(_xyz(dst_lon, dst_lat), workers=-1)
    covered = dist <= cutoff

    out = {n: np.where(covered, ds_src[n].values.ravel()[src_land][keep][j], np.nan)
           for n in names}
    return out, covered


def fill_nearest(vals, lon, lat, fill_mask):
    '''Fill from the nearest valid column -- `sit.fast_interpolate`, method='nearest'.'''
    need = fill_mask & ~np.isfinite(vals)
    if not need.any():
        return vals
    good = np.isfinite(vals)
    _, j = cKDTree(_xyz(lon[good], lat[good])).query(
        _xyz(lon[need], lat[need]), workers=-1)
    out = vals.copy()
    out[need] = vals[good][j]
    return out


# --- kriging gap-fill: `sit.slow_interpolate`, made tractable ------------------

def fit_variogram(vals, lon, lat, n_patches=40, spacing=100,
                  variogram_model='spherical', seed=0):
    '''Fit the variogram ONCE, at the lag scale the kriging actually uses.

    `sit.slow_interpolate` -- and the first version of this notebook -- refits a
    variogram inside every solve, i.e. tens of thousands of independent fits from
    100 points each. That is ~95% of the runtime and it is *worse*: a 100-point
    fit is badly under-determined, and on held-out columns the per-point scheme
    scored MAE 56.2 m at ne30np4 against 50.6 m for the pooled fit below, barely
    beating plain nearest-neighbour (59.0 m).

    Fitting on a *global* sample does not work either -- those lags all sit far
    beyond the correlation length, the fit collapses to pure nugget (range -> 0)
    and MAE roughly doubles. Pooling real neighbourhood patches is what keeps the
    lag range honest.

    Returns a dict keyed `'psill'`, deliberately: pykrige treats a bare list and
    the `'sill'` key as a *total* sill and subtracts the nugget, so feeding back
    `variogram_model_parameters` (which is [psill, range, nugget]) through either
    of those silently shrinks the partial sill.
    '''
    valid = np.flatnonzero(np.isfinite(vals))
    tree = cKDTree(_xyz(lon[valid], lat[valid]))
    rng = np.random.default_rng(seed)
    params = []
    for i in rng.choice(valid, min(n_patches, valid.size), replace=False):
        _, nb = tree.query(_xyz(lon[i], lat[i]), k=spacing)
        dlon = ((lon[valid][nb] - lon[i] + 180.0) % 360.0) - 180.0
        dlat = lat[valid][nb] - lat[i]
        try:
            uk = UniversalKriging(dlon, dlat, vals[valid][nb],
                                  variogram_model=variogram_model,
                                  drift_terms=['linear'], verbose=False,
                                  enable_plotting=False)
            params.append(uk.variogram_model_parameters)
        except Exception:
            pass
    psill, range_, nugget = np.median(np.array(params), axis=0)
    return {'psill': psill, 'range': range_, 'nugget': nugget}


# Workers share the KD-tree and every field's values through this module-level
# dict, populated once per worker by `_krig_init`.
_KRIG = {}


def _krig_init(vxyz, vlon, vlat, vvals, vparams, variogram_model, spacing):
    _KRIG.update(tree=cKDTree(vxyz), vlon=vlon, vlat=vlat, vvals=vvals,
                 vparams=vparams, vgm=variogram_model, k=spacing)


def _krig_solve(args):
    '''Solve every field at one gap column, sharing the neighbour lookup.'''
    tlon_i, tlat_i, txyz = args
    g = _KRIG
    # One query serves all fields: they come from the same rotation, so their
    # valid masks -- and therefore their neighbour sets -- are identical.
    _, nb = g['tree'].query(txyz, k=g['k'], workers=1)
    # Krige in a frame centred on the target column: this unwraps the longitude
    # seam and keeps the linear drift term well conditioned at high latitude.
    dlon = ((g['vlon'][nb] - tlon_i + 180.0) % 360.0) - 180.0
    dlat = g['vlat'][nb] - tlat_i
    out = []
    for name, vals in g['vvals'].items():
        val = vals[nb]
        try:
            kr = UniversalKriging(dlon, dlat, val, variogram_model=g['vgm'],
                                  drift_terms=['linear'],
                                  variogram_parameters=g['vparams'][name],
                                  verbose=False, enable_plotting=False)
            out.append(float(kr.execute('points', np.array([0.0]), np.array([0.0]))[0][0]))
        except Exception:
            out.append(float(val[0]))  # degenerate neighbourhood -> nearest valid
    return out


def fill_kriging(fields, lon, lat, fill_mask, variogram_model='spherical',
                 spacing=None, workers=16, chunk=64, verbose=True):
    '''Kriging gap-fill for several fields at once.

    `fields` is {name: 1-D array}. All of them must share the same valid mask --
    true here because they come from one rotation -- which is what lets a single
    KD-tree and a single neighbour lookup serve every field.

    Deviations from `sit.slow_interpolate`, both measured:
      * the variogram is fitted once per field (see `fit_variogram`) instead of
        once per gap column: ~2x faster and ~10% lower MAE;
      * `spacing` neighbours come from a KD-tree rather than an O(N) argsort per
        point, and `spacing=50` rather than 100 is ~2.5x faster at equal accuracy.
    Together ~5x against the per-point scheme, and more accurate.

    The kriging itself is untouched: universal kriging, spherical variogram,
    linear drift, in the target-centred frame.
    '''
    spacing = krig_spacing if spacing is None else spacing
    names = list(fields)
    valid = np.isfinite(fields[names[0]])
    for n in names[1:]:
        assert np.array_equal(np.isfinite(fields[n]), valid), \
            f'{n} has a different valid mask; the shared neighbour lookup assumes one mask'
    need = fill_mask & ~valid
    if not need.any():
        return dict(fields)

    t0 = time.time()
    params = {n: fit_variogram(fields[n], lon, lat) for n in names}
    if verbose:
        for n in names:
            p = params[n]
            print(f'    {n} variogram: psill={p["psill"]:.0f} '
                  f'range={p["range"]:.3f} deg nugget={p["nugget"]:.0f}')
        print(f'    fitted {len(names)} variograms in {time.time() - t0:.0f} s; '
              f'{int(need.sum())} gap columns x {len(names)} fields')

    args = list(zip(lon[need], lat[need], _xyz(lon[need], lat[need])))
    with ProcessPoolExecutor(
        workers, initializer=_krig_init,
        initargs=(_xyz(lon[valid], lat[valid]), lon[valid], lat[valid],
                  {n: fields[n][valid] for n in names}, params,
                  variogram_model, spacing),
    ) as ex:
        rows = list(ex.map(_krig_solve, args, chunksize=chunk))

    rows = np.asarray(rows)
    out = {}
    for c, n in enumerate(names):
        a = fields[n].copy()
        a[need] = rows[:, c]
        out[n] = a
    return out
```

```python
%%time

# Each target rotates its *own* reference topo, straight from ncol to ncol. That
# restores the resolution matching 5.1 has -- SGH grows with cell size, so
# ne30np4 needs ne30np4 roughness, not ne120np4's.
FILLS = ['raw', 'nn', 'krig']  # unfilled / nearest / kriging
ROT_VARS = ['SGH', 'SGH30']

for tgt, cfg in targets.items():
    for f in FILLS:
        cfg[f'sgh_rot_se_{f}'] = os.path.join(
            output_dir, f'sgh_rot_{f}_{case}_{tgt}.{today}.nc')
    cfg['topo_out_rot'] = os.path.join(
        output_dir, f'cami_bnd_topo_{tgt}_{case}_rotsgh.{today}.nc')

for tgt, cfg in targets.items():
    t0 = time.time()
    print(f'=== {tgt} ===')
    ds_src = xr.open_dataset(cfg['ref_topo'])
    ds_dst = xr.open_dataset(cfg['topo_out'])          # 5.1 output: the Miocene grid
    dlon, dlat = column_lonlat(ds_dst)                 # flat columns for SE and FV alike
    mio_land = ds_dst['LANDFRAC'].values.ravel() > 0

    radius = gap_radius_factor * grid_spacing_deg(dlon, dlat)
    # Mask to present-day land before rotating: ocean columns carry no meaningful
    # roughness and would smear onto Miocene continents. One rotation call covers
    # both fields, since what moves is the column coordinates.
    rot, covered = rotate_columns(
        ds_src, ROT_VARS, ds_src['LANDFRAC'].values > 0, dlon, dlat, radius)

    n_gap = int((mio_land & ~covered).sum())
    print(f'  rotated in {time.time() - t0:.0f} s; gap radius {radius:.2f} deg; '
          f'{n_gap} / {int(mio_land.sum())} Miocene-land columns '
          f'({100 * n_gap / mio_land.sum():.1f}%) had no present-day parcel land on them')

    variants = {'raw': rot,
                'nn': {n: fill_nearest(rot[n], dlon, dlat, mio_land) for n in ROT_VARS}}
    # Both fields in one pass: same gap columns, same neighbour sets, so the
    # KD-tree and the per-column lookup are built once and reused.
    t1 = time.time()
    variants['krig'] = fill_kriging(rot, dlon, dlat, mio_land)
    print(f'  kriging {ROT_VARS} together: {time.time() - t1:.0f} s')

    # Keep Miocene land only; ocean is set in the assembly cell below.
    # Written in the target's native layout -- flat ncol for SE, (lat, lon) for
    # FV -- so the assembly and comparison cells treat the rotated files exactly
    # like the 5.1 outputs.
    tmpl = ds_dst['LANDFRAC']
    for f in FILLS:
        ds_rot = xr.Dataset(
            {f'{n}_rot': (tmpl.dims,
                          np.where(mio_land, variants[f][n], np.nan).reshape(tmpl.shape))
             for n in ROT_VARS})
        if tmpl.ndim == 1:  # SE: per-column coords
            ds_rot = ds_rot.assign_coords(lat=('ncol', dlat), lon=('ncol', dlon))
        else:               # FV: 1-D axis coords
            ds_rot = ds_rot.assign_coords(lat=ds_dst['lat'], lon=ds_dst['lon'])
        for v in ds_rot.data_vars.values():
            v.encoding['_FillValue'] = 1.e20
        ds_rot.to_netcdf(cfg[f'sgh_rot_se_{f}'],
                         encoding={c: {'_FillValue': None} for c in ['lat', 'lon']})
    print(f'  --> {cfg["sgh_rot_se_krig"]}  ({time.time() - t0:.0f} s total)')
```

```python
%%time
glon_range, glat_range = [-180, 180], [-90, 90]

rows = [
    ('present-day SGH30 (ref_topo)', {tgt: xr.open_dataset(cfg['ref_topo']) for tgt, cfg in targets.items()}),
    ('regression on PHIS', {tgt: xr.open_dataset(cfg['topo_out']) for tgt, cfg in targets.items()}),
    ('rotated (gaps)', {tgt: xr.open_dataset(cfg['sgh_rot_se_raw']) for tgt, cfg in targets.items()}),
    ('nearest-filled', {tgt: xr.open_dataset(cfg['sgh_rot_se_nn']) for tgt, cfg in targets.items()}),
    ('kriging-filled', {tgt: xr.open_dataset(cfg['sgh_rot_se_krig']) for tgt, cfg in targets.items()}),
]

# Miocene land mask per target, for the ocean -> 0 rule on the 5.2 rows.
mio_land_of = {tgt: xr.open_dataset(cfg['topo_out'])['LANDFRAC'].values > 0 for tgt, cfg in targets.items()}

sgh_kws = dict(
    cmap=plt.cm.terrain, cmin=0, cmax=400,
    lon_range=glon_range, lat_range=glat_range,
    lonlat_line=True, lon_interval=60, lat_interval=30,
    colorbar=False, nticks=9, label_size=8, title_size=10,
    coast=False, country=False, state=False,
)

fig, axes = plt.subplots(
    5, len(targets), figsize=(5 * len(targets), 13.5), constrained_layout=True,
    subplot_kw={'projection': ccrs.PlateCarree()},
)

# ne120np4 is ~777k polygons per panel, so this takes a few minutes.
for r, (label, se_by_tgt) in enumerate(rows):
    kw = dict(sgh_kws, coast=(r == 0))   # modern coastline only on the present-day row
    for c, (tgt, ds_se) in enumerate(se_by_tgt.items()):
        var = 'SGH30' if 'SGH30' in ds_se else 'SGH30_rot'
        vals = ds_se[var].values
        if var == 'SGH30_rot':             # 5.2 intermediate -> apply the shipped ocean rule
            vals = np.where(mio_land_of[tgt], vals, 0.0)
        if vals.ndim == 2:   # FV target: explicit axes, no scripgrid needed
            p = Plot_2D(vals, lons=ds_se['lon'].values, lats=ds_se['lat'].values,
                        ax=axes[r, c], title=f'{tgt}: {label}', **kw)
        else:                # SE target: element corners from the scripgrid
            p = Plot_2D(vals, ax=axes[r, c], scrip_file=targets[tgt]['atm_scrip'],
                        title=f'{tgt}: {label}', **kw)

# The 15 Ma shoreline on the palaeo rows. It now does real work: with the ocean
# filled, this is what separates 0 m ocean from genuinely smooth land.
_lon = ((ds_raw025['lon'].values + 180) % 360) - 180
_o = np.argsort(_lon)
for ax in axes[1:].ravel():
    ax.contour(_lon[_o], ds_raw025['lat'].values, ds_raw025['landmask'].values[:, _o], levels=[0.5], colors='k', linewidths=0.4, transform=ccrs.PlateCarree())

cbar_ax = fig.add_axes([0.15, -0.025, 0.70, 0.012])
fig.colorbar(
    p.im, cax=cbar_ax, extend='both', orientation='horizontal',
    ticks=p.colorticks, label='SGH30 [m]',
)

for tgt, cfg in targets.items():
    land = mio_land_of[tgt]
    nn = xr.open_dataset(cfg['sgh_rot_se_nn'])['SGH30_rot'].values
    kg = xr.open_dataset(cfg['sgh_rot_se_krig'])['SGH30_rot'].values
    d = np.abs(nn - kg)[np.isfinite(nn) & np.isfinite(kg)]
    neg = land & np.isfinite(kg) & (kg < 0)
    print(f'{tgt}: |nearest - kriging| over land  mean {d.mean():.1f} m,  max {d.max():.1f} m')
    print(f'   kriging returns SGH30 < 0 on {int(neg.sum())} land columns '
          f'(min {kg[neg].min():.1f} m)' if neg.any() else
          f'   kriging returns no negative SGH30')
```
